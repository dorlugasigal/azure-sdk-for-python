# Azure AI Evaluation SDK v2.0 — POC

> evee engine + real SDK evaluators. Local or cloud. Same `evaluate_v2()` call.


```python
import os, sys, json
from IPython.display import display, HTML

if os.path.basename(os.getcwd()) == "demo":
    os.chdir("..")
if not os.path.exists("demo/data.jsonl"):
    for p in ["/Users/dorlugasigal/Projects/azure-sdk-for-python/sdk/evaluation/azure-ai-evaluation", os.path.join(os.getcwd(), "..")]:
        if os.path.exists(os.path.join(p, "demo", "data.jsonl")):
            os.chdir(p)
            break

from azure.ai.evaluation._eval_v2 import evaluate_v2
from azure.ai.evaluation import RelevanceEvaluator, F1ScoreEvaluator
from azure.identity import DefaultAzureCredential

MODEL = {"azure_deployment": "gpt-4.1-mini", "azure_endpoint": "https://foundry-evee-ko9z2s7c.cognitiveservices.azure.com/", "type": "azure_openai"}
CRED = DefaultAzureCredential()
DATA = "demo/data.jsonl"

def results_table(result, title="Results"):
    rows_html = ""
    for i, row in enumerate(result["rows"]):
        q = row["input"].get("question", "")[:40]
        score_cells = ""
        for name, scores in row["scores"].items():
            if isinstance(scores, dict) and "error" not in scores:
                val = next((v for k, v in scores.items() if isinstance(v, (int, float)) and "threshold" not in k and "tokens" not in k), "?")
                score_cells += f"<td style='text-align:center'>{val}</td>"
            else:
                score_cells += f"<td style='text-align:center; color:red'>err</td>"
        rows_html += f"<tr><td>{i+1}</td><td>{q}</td>{score_cells}</tr>"
    headers = "".join(f"<th>{n}</th>" for n in result["rows"][0]["scores"].keys())
    metrics_html = " | ".join(f"<b>{k}</b>: {v}" for k, v in result["metrics"].items())
    display(HTML(f"<h4>{title}</h4><table border=1 cellpadding=4 style='border-collapse:collapse'><tr><th>#</th><th>Question</th>{headers}</tr>{rows_html}</table><p>{metrics_html}</p>"))

print(f"Ready — {os.getcwd()}")
```

## 1. Local Evaluation — No Model Needed

F1 score is a deterministic NLP metric ported from the current SDK. It compares token overlap between the response and ground truth — runs instantly, fully offline, no model or API key required.


```python
result_local = evaluate_v2(
    data=DATA,
    evaluators={"f1": F1ScoreEvaluator()},
    evaluator_config={"f1": {"column_mapping": {"response": "${data.answer}", "ground_truth": "${data.context}"}}},
)
results_table(result_local, "Local: F1 Score (no model, instant)")
```

## 2. Cloud Evaluation — GPT-4.1-mini via Foundry

Same `evaluate_v2()` function, but the evaluator is initialized with a `model_config` pointing to our Foundry deployment. The **real** Azure AI Evaluation SDK's `RelevanceEvaluator` calls GPT-4.1-mini as an LLM-as-judge — returning a score, reasoning, model name, and token usage.


```python
result_cloud = evaluate_v2(
    data=DATA,
    evaluators={"relevance": RelevanceEvaluator(model_config=MODEL, credential=CRED)},
    evaluator_config={"relevance": {"column_mapping": {"query": "${data.question}", "response": "${data.answer}"}}},
)
results_table(result_cloud, "Cloud: Relevance via GPT-4.1-mini (LLM-as-judge)")
```

## 3. Local + Cloud Together

Mix evaluator types in a single call: F1 (local, instant, deterministic) alongside Relevance (cloud, LLM-as-judge). Both produce structured results in the same output.


```python
result_both = evaluate_v2(
    data=DATA,
    evaluators={
        "f1": F1ScoreEvaluator(),
        "relevance": RelevanceEvaluator(model_config=MODEL, credential=CRED),
    },
    evaluator_config={
        "f1": {"column_mapping": {"response": "${data.answer}", "ground_truth": "${data.context}"}},
        "relevance": {"column_mapping": {"query": "${data.question}", "response": "${data.answer}"}},
    },
)
results_table(result_both, "Combined: F1 (local) + Relevance (cloud)")
```

## 4. Config-Driven via YAML

Instead of writing Python, define the entire evaluation in `evals.yaml` — dataset, metrics, model endpoint, parameters. The evee engine reads the config, discovers the registered metrics, creates the real SDK evaluators with the connection details, and runs everything automatically.


```python
with open("demo/evals.yaml") as f:
    display(HTML(f"<pre>{f.read()}</pre>"))
```

```python
result_config = evaluate_v2(config="demo/evals.yaml")
display(HTML(f"<b>Status:</b> {result_config['status']} | <b>Records:</b> {result_config['total_records']} | <b>Variants:</b> {result_config['models_evaluated']}"))
```

### Output Files

The engine persists results to disk — a JSONL file with per-record scores and a JSON summary with aggregated metrics. These files are the evaluation record: reproducible, diffable, and ready for upload to Foundry.


```python
from pathlib import Path
import shutil

output_dir = Path(result_config["output_path"])
print(f"Output directory: {output_dir}\n")

for f in sorted(output_dir.iterdir()):
    size = f.stat().st_size
    print(f"  {f.name:<40} {size:>6} bytes")

# Show the summary
for f in output_dir.glob("*_summary.json"):
    print(f"\n{'='*60}")
    print(f"  {f.name}")
    print(f"{'='*60}")
    with open(f) as fh:
        summary = json.load(fh)
    print(f"  Model:   {summary['model']}")
    print(f"  Records: {summary['total_records']}")
    print(f"  Metrics:")
    for metric_name, metric_data in summary.get("aggregated_metrics", {}).items():
        for k, v in metric_data.items():
            print(f"    {k}: {v}")

# Show first record from JSONL
for f in output_dir.glob("*_results.jsonl"):
    print(f"\n{'='*60}")
    print(f"  {f.name} (first record)")
    print(f"{'='*60}")
    with open(f) as fh:
        first = json.loads(fh.readline())
    metrics = first.get("metrics", {})
    for metric_name, scores in metrics.items():
        relevant_keys = {k: v for k, v in scores.items() if not k.endswith("_sample_input") and not k.endswith("_sample_output")}
        print(f"  {metric_name}:")
        for k, v in relevant_keys.items():
            print(f"    {k}: {v}")
    break

# Cleanup
shutil.rmtree(output_dir.parent.parent, ignore_errors=True)
```

## 5. CLI — Run + View Results

The same config-driven evaluation runs from the terminal. The CLI reads `evals.yaml`, invokes the engine, and writes results to the output directory. Then we inspect what it produced.


```python
# Run evaluation from CLI
import subprocess, re

proc = subprocess.run(
    ["python", "-m", "azure.ai.evaluation.cli", "run", "-c", "demo/evals.yaml"],
    capture_output=True, text=True
)
print(proc.stdout)

# Capture the output path
match = re.search(r"Output:\s+(.+)", proc.stdout)
cli_output_path = match.group(1).strip() if match else None
print(f"Output path: {cli_output_path}")
```

```python
# View the results from CLI run
if cli_output_path:
    from pathlib import Path

    output_dir = Path(cli_output_path)
    print(f"Files in {output_dir.name}/\n")
    for f in sorted(output_dir.iterdir()):
        print(f"  {f.name:<45} {f.stat().st_size:>6} bytes")

    # Summary
    for f in output_dir.glob("*_summary.json"):
        with open(f) as fh:
            summary = json.load(fh)
        print(f"\nSummary: {summary['model']}")
        for mn, md_data in summary.get("aggregated_metrics", {}).items():
            for k, v in md_data.items():
                print(f"  {k}: {v}")

    # First record
    for f in output_dir.glob("*_results.jsonl"):
        with open(f) as fh:
            first = json.loads(fh.readline())
        print(f"\nFirst record:")
        for mn, scores in first.get("metrics", {}).items():
            score = scores.get(mn, "?")
            reason = scores.get(f"{mn}_reason", "")
            model = scores.get(f"{mn}_model", "")
            print(f"  {mn}: {score}")
            if reason:
                print(f"  reason: {reason[:100]}")
            if model:
                print(f"  model:  {model}")
        break

    # Cleanup
    import shutil
    shutil.rmtree(output_dir.parent.parent, ignore_errors=True)
```

## Summary

| Mode | Code | What Happens |
|------|------|-------------|
| **Local** | `evaluate_v2(data=..., evaluators={"f1": F1ScoreEvaluator()})` | Instant, no model, offline |
| **Cloud** | `evaluate_v2(data=..., evaluators={"rel": RelevanceEvaluator(model_config=MODEL)})` | GPT-4.1-mini judge via Foundry |
| **Config** | `evaluate_v2(config="evals.yaml")` | YAML drives everything |
| **CLI** | `python -m azure.ai.evaluation.cli run -c evals.yaml` | Same from terminal |

**Same evaluators. Same data. Local or cloud. Config or code or CLI.**
